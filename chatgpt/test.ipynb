```python
# Absolute top level (Global Scope)
IS_MAINTENANCE_MODE = False

def boot_user_session():
    # Reads the global variable safely without needing the keyword
    if IS_MAINTENANCE_MODE:
        return "Error 503: Site under maintenance."
    return "Welcome back, user!"

def emergency_shutdown():
    global IS_MAINTENANCE_MODE  # 🚨 Reaches out to the top level
    IS_MAINTENANCE_MODE = True  # Flips the switch for the whole application
    print("Emergency shutdown activated system-wide!")

# --- Simulation ---
print(boot_user_session()) # Output: Welcome back, user!

emergency_shutdown()       # This function changes the global variable

print(boot_user_session()) # Output: Error 503: Site under maintenance.
```
