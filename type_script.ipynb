```python
class Razorpay:
    def pay(self, amount: float) -> None:
        print("Paid with Razorpay ", amount)


class Stripe:
    def pay(self, amount: float) -> None:
        print("Paid with Stripe")
```

```python
def checkout(payment_system, amount: float) -> None:
    payment_system.pay(amount)
```

```python
a = Razorpay()
a.pay(1000)
```

```python
checkout(payment_system=Razorpay,1000.0)
```
